Mode 
the mode value is the value that appears the most number of times:

Use the SciPy mode() method to find the number that appears the most:

Standard deviation

standard deviation is a number that describes how spread out the values are.

A low standard deviation means that most of the numbers are close to the mean (average) value.

A high standard deviation means that the values are spread out over a wider range.

Example : this time we have registered the speed of 7 cars:

speed = [86, 87, 88, 86, 87, 85, 86]

the standard deviation is : 0.9

meaning that most of the values are within the range of 0.9 from the mean value, whic is 86.4. 

speed = [32, 111, 138, 28, 59, 77, 97]

the standard deviation is:

37.85

meaning that most if the values are within the range of 37.85 from the mean value, which is 77.4

as you can see, a higher standard deviation indicates that the values are spread out over a wider range.

the NumPy module has a method to calculate the standard deviation:

```python
#use the NumPy std() method to find the standard deviation:
import numpy as np
speed = [86, 87, 88, 86, 87, 85, 86]
x = np.std(speed)
print(x)
```

```python
import numpy as np
speed = [32, 111, 138, 28, 59, 77, 97]
x = np.std(speed)
print(x)
```

Variance

Variance is another number that indicates how spread out the values are.

In fact, if you take the square root of the variance, you get the standard deviation!

Or the other way around, if you multiply the standard deviation by itself, you get the variance!

To calculate the variance you have to do as follows:
 1. Find the mean
 2. For each value : find the difference from the mean:
 3. For each diffrence: find the square value:
 4. The variance is the average number of these squared fifferences:

 [link](https://www.w3schools.com/python/python_ml_standard_deviation.asp)



```python
#use the NumPy var() method to find the variance:
import numpy as np
speed = [32, 111, 138, 28, 59, 77, 97]
x = np.var(speed)
print(x)
```

```python
#Use the NumPy std() method to find the standard deviation, and the NumPy var() method to find the variance:
import numpy as np
speed = [32, 111, 138, 28, 59, 77, 97]
x = np.std(speed)
print(x)
```

Percentiles

Percentiles are used in statistics to give you a number that describes the values that given percent of the values are lower than.

ex: Let's say we have an array that contains the ages of every person living on a street

ages = [5,31,43,48,50,41,7,11,15,39,80,82,32,2,8,6,25,36,27,61,31]

What is the 75. percentile? 
The answer is 43, meaning that 75% of the people are 43 or younger.
The NumPy module has a method for finding the specific percentile.


```python
#Use the Numpy percentile() method to find the percentile of an array:
import numpy as np
ages = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80, 82, 32, 2, 8, 6, 25, 36, 27, 61, 31]
x = np.percentile(ages, 75)
print(x)
```

```python
#what is the age that 90% of the people are younger than?
import numpy as np
ages = [5, 31, 43, 48, 50, 41, 7, 11, 15, 39, 80, 82, 32, 2, 8, 6, 25, 36, 27, 61, 31]
x = np.percentile(ages, 90)
print(x)
```
