#01 - Tensor Basics

Goal : create , inspect, and calculate with TensorFlow tensors.

```python
import tensorflow as tf
print("TensorFlow version:", tf.__version__)
```

```python
my_arr = [[1, 2, 3, 4]]
tensor_a = tf.constant(my_arr)

tensor_a
```

```python
print("Shape:", tensor_a.shape)
print("Data type:", tensor_a.dtype)
print("As Numpy:", tensor_a.numpy())
```

```python
tensor_b = tf.constant([[10, 20, 30, 40]])

tensor_a + tensor_b
```
