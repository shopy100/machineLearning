```python
import requests

# Define a custom User-Agent header to mimic a web browser
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# Make the request with the custom header
response = requests.get(URL, headers=headers)

# Check if the request was successful
if response.status_code != 200:
    print(f"Failed to retrieve content. Status code: {response.status_code}")
    # Exit if content retrieval failed
    raise ConnectionError(f"Failed to retrieve content. Status code: {response.status_code}")
```
